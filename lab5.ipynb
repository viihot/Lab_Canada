**3. Реалізація алгоритмічного завдання**

```python
import math

def calculate_factorial(n):
    # перевірка, чи є введене значення цілим числом
    if not isinstance(n, int):
        raise TypeError("помилка: введіть ціле число.")
    # перевірка на невід'ємність
    if n < 0:
        raise ValueError("помилка: число має бути невід'ємним.")

    return math.factorial(n)


число = 5
результат = calculate_factorial(число)
print(f"факторіал числа {число} дорівнює {результат}")
```

**4. Тестові приклади**

```python
test_values = [0, 5, 7]
for val in test_values:
    res = calculate_factorial(val)
    print(f"факторіал {val} = {res}")
```
