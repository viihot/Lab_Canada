```python
def find_median(data_list):
    # Контроль правильності введення даних [cite: 29]
    if not isinstance(data_list, list):
        return "Помилка: вхідні дані мають бути списком."
    if not data_list:
        return "Помилка: список порожній."
    if not all(isinstance(x, (int, float)) for x in data_list):
        return "Помилка: список повинен містити лише числа."

    # Алгоритм знаходження медіани
    n = len(data_list)
    sorted_data = sorted(data_list)
    mid = n // 2

    if n % 2 == 0:
        # Для парної кількості елементів — середнє двох середніх
        median = (sorted_data[mid - 1] + sorted_data[mid]) / 2
    else:
        # Для непарної — центральний елемент
        median = sorted_data[mid]

    return median
```

```python
# Тестові приклади
test_cases = [
    [1, 3, 5],             # Непарна кількість
    [10, 2, 5, 8],         # Парна кількість
    [7, 1, 9, 3, 5, 11]    # Ще один приклад
]

for i, test in enumerate(test_cases, 1):
    result = find_median(test)
    print(f"Тест {i} (список {test}): Медіана = {result}")

# Приклад перевірки неправильного введення [cite: 29]
print("Перевірка помилки:", find_median("не список"))
```
